```python
from watermark_detector import UNGATED_GEMMA_TOKENIZER, SynthIDDetector

# tokenizer_repo points at an ungated mirror of google/gemma-2b-it, whose tokenizer the
# detector needs. Drop the argument if you have accepted the Gemma licence on the Hub.
detector = SynthIDDetector(tokenizer_repo=UNGATED_GEMMA_TOKENIZER)

WATERMARKED_TEXT = """
## Building the Beacon of Hope: The Evolution of Lighthouse Construction and Operation

The lighthouse, a sentinel on the vast seas, rose from the need for safe passage through treacherous waters. Construction methods and technologies revolutionized with each technological leap, building lighthouses that continue to withstand the fury of nature.

Wooden lighthouses, the first pioneers, were sturdy structures, handcrafted with local materials and fueled by coal or whale oil lamps. These beacons, often square-shaped with a circular light, were surprisingly effective, casting a beam up to 10 miles, aiding maritime travelers and commercial vessels. However, their wooden frames were susceptible to erosion and fire, necessitating regular maintenance and replacement.

The iron lighthouse, a steel marvel, emerged in the late 19th century. Made of iron sheets, this sturdy structure was more fire-resistant and could withstand storms. It boasted a large lens, offering a far greater range and brighter light than its wooden predecessor. While the iron lighthouse was initially more expensive, its longevity and resilience made it a beacon of hope in the vast ocean.

The 20th century brought significant advancements. Cement and steel replaced wood and iron, creating lighthouses with improved structural integrity and weather resistance. Coastal communities began utilizing oil lamps, replacing the more volatile whale oil, and installing automated mechanisms to maintain lamp brightness and extend battery life. These lighthouses, the epitome of human ingenuity and the dawn of a new era in maritime navigation, continue to stand as enduring symbols of human resilience and technological progress.
"""

result = detector.detect(WATERMARKED_TEXT)

# score is the Bayesian posterior and drives the verdict. z_score is the simpler
# frequentist view: how many standard deviations the mean g-value sits above the 0.5 a
# fair coin would give. Unwatermarked text lands near 0, roughly +-2 either way.
print(f"verdict:        {result.verdict}")
print(f"score:          {result.score:.4f}")
print(f"is_watermarked: {result.is_watermarked}")
print(f"z_score:        {result.z_score:+.2f}")
print(f"mean g-value:   {result.mean_g_value:.4f}  (0.5 = no watermark)")
print(f"tokens scored:  {result.token_count}")
print(f"reliable:       {result.reliable}")
```
